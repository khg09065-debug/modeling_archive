```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine
from sklearn.cluster import KMeans, OPTICS  # 📌 HDBSCAN 대신 OPTICS 모델을 임포트합니다.
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# ==========================================
# 1. 사용자 설정 (K값 및 가성비/프리미엄 브랜드 가중치)
# ==========================================
n_clusters = 4  # K-Means 군집의 개수를 4개로 설정합니다.

# 브랜드별 가성비(value) 및 프리미엄(premium) 점수를 정의합니다.
brand_weights = {
    '메머드커피':  {'value': 1, 'premium': 1.6},
    '컴포즈커피':  {'value': 1, 'premium': 2},
    '더벤티':      {'value': 1, 'premium': 2},
    '하삼동커피':  {'value': 1, 'premium': 2},
    '메가커피':    {'value': 1, 'premium': 2.2},
    '빽다방':      {'value': 1, 'premium': 2.2},
    '디저트39':    {'value': 1, 'premium': 3},
    '이디야':      {'value': 1, 'premium': 3.2},
    '던킨도너츠':  {'value': 1, 'premium': 4},
    '할리스':      {'value': 1, 'premium': 4.5},
    '공차':        {'value': 1, 'premium': 5},
    '파스쿠찌':    {'value': 1, 'premium': 5},
    '투썸플레이스': {'value': 1, 'premium': 5},
    '스타벅스':    {'value': 1, 'premium': 5},
    '폴바셋':      {'value': 1, 'premium': 5}
}

# --- [그래프 폰트 환경 설정] ---
plt.rcParams['font.family'] = 'Malgun Gothic'  # 한글 깨짐 방지를 위해 맑은 고딕 폰트를 설정합니다.
plt.rcParams['axes.unicode_minus'] = False     # 그래프에서 마이너스 기호가 깨지는 현상을 방지합니다.

# ==========================================
# 2. 데이터 로드 및 지역 필터링 (서울 & 경기도)
# ==========================================
# 커피 매장 데이터베이스와 지하철 역사 데이터베이스 엔진을 각각 생성합니다.
coffee_engine = create_engine('mysql+pymysql://root:1234@localhost:3306/coffee_store')
subway_engine = create_engine('mysql+pymysql://root:1234@localhost:3306/subway')

# subway DB의 total_subway 테이블에서 데이터를 직접 로드합니다.
raw_df = pd.read_sql("SELECT * FROM total_subway", subway_engine)

# 서울과 경기도만 필터링
raw_df = raw_df[raw_df['지역'].isin(['서울', '경기'])].copy()
# 1월 승차객수와 하차객수를 더하여 '총_승하차객수' 파생변수를 생성합니다.
raw_df['총_승하차객수'] = raw_df['1월 승차이용객수'] + raw_df['1월 하차이용객수']

# 지역과 역명을 기준으로 그룹화하여 환승역의 승하차객수를 합산하고 위경도 평균값을 계산합니다.
station_base = raw_df.groupby(['지역', '역명']).agg({
    '총_승하차객수': 'sum', 
    '위도': 'mean', 
    '경도': 'mean'
}).reset_index()

# OPTICS 거리 계산(Haversine)을 위해 위도와 경도 좌표를 라디안 단위로 변환합니다.
coords = np.radians(station_base[['위도', '경도']].values)

# 📌 [수정 사항] HDBSCAN 대신 밀도 편차 조율이 가능한 OPTICS 모델을 초기화합니다.
# 서울과 경기 지역의 지하철역 밀도 차이를 유연하게 반영하기 위해 'xi' 추출 기법을 활용합니다.
optics = OPTICS(
    min_samples=2, 
    max_eps=float(1/6371),
    metric='haversine',
    cluster_method='xi',     # 밀도 변화 기울기를 감지하는 xi 방식으로 설정합니다.
    xi=0.05,                 # 군집 구분을 짓는 임계 경사값
    min_cluster_size=2       # 상권으로 묶일 최소 역 개수
)
# 라디안 좌표를 기반으로 OPTICS 연산을 수행하고, 각 전철역에 상권_ID를 부여합니다.
station_base['상권_ID'] = optics.fit_predict(coords)

# 외딴 독립 역(-1)들이 서로 다른 지역임에도 하나로 뭉치는 문제를 방지하기 위해 개별 고유 ID를 부여합니다.
max_id = station_base['상권_ID'].max()
noise_mask = (station_base['상권_ID'] == -1)
num_noise = noise_mask.sum()
station_base.loc[noise_mask, '상권_ID'] = range(max_id + 1, max_id + 1 + num_noise)

# 동일한 상권_ID를 가진 역세권들을 묶어서 상권 중심점(위경도 평균)과 총 유동인구를 구합니다.
districts = station_base.groupby('상권_ID').agg({
    '역명': lambda x: ', '.join(x.unique()),
    '지역': 'first', 
    '총_승하차객수': 'sum', 
    '위도': 'mean', 
    '경도': 'mean'
}).reset_index()

# MySQL 데이터베이스에서 전체 커피 체인 매장 데이터를 쿼리하여 로드합니다.
coffee_df_raw = pd.read_sql("SELECT 브랜드명, 매장명, 경도, 위도, 지역 FROM coffee_chain", coffee_engine)
# 분석 중인 상권 지역(서울, 경기)에 존재하는 커피 매장들만 필터링합니다.
valid_regions = districts['지역'].unique().tolist()
coffee_df = coffee_df_raw[coffee_df_raw['지역'].isin(valid_regions)].copy()

# ==========================================
# 3. 가성비 & 프리미엄 가중치 통합 변수 계산
# ==========================================
def get_weighted_features(row, stores):
    # 각 상권 중심점의 위도와 경도를 가져옵니다.
    s_lat, s_lon = row['위도'], row['경도']
    # 중심점 기준 일정 범위 내에 존재하는 주변 커피 매장들을 1차 필터링합니다.
    nearby = stores[(stores['위도'].between(s_lat - 0.005, s_lat + 0.005)) & 
                    (stores['경도'].between(s_lon - 0.006, s_lon + 0.006))]
    
    # 주변에 매장이 없으면 전체 매장수 0, 가성비 지수 0, 프리미엄 지수 0을 반환합니다.
    total_count = len(nearby)
    if total_count == 0:
        return pd.Series([0, 0, 0])
    
    value_sum = 0
    premium_sum = 0
    # 반경 내 탐색된 매장들의 브랜드를 체크하여 가성비 및 프리미엄 점수를 각각 누적 합산합니다.
    for brand in nearby['브랜드명']:
        if brand in brand_weights:
            value_sum += brand_weights[brand]['value']
            premium_sum += brand_weights[brand]['premium']
            
    # 누적된 합산 점수를 기반으로 최종 지수화(두 피처 모두 SUM 활용)를 진행합니다.
    weighted_b_index = value_sum
    premium_b_index = premium_sum
    
    return pd.Series([total_count, round(weighted_b_index, 2), round(premium_b_index, 2)])

print("📊 브랜드별 정수 가중치 적용 및 상권 통합 평가지표 분석 중...")
# 모든 상권에 대해 위 함수를 적용하여 '전체_매장수', '가성비_지수', '프리미엄_지수' 컬럼을 추가합니다.
districts[['전체_매장수', '가성비_지수', '프리미엄_지수']] = districts.apply(
    lambda x: get_weighted_features(x, coffee_df), axis=1
)

# ==========================================
# 4. K-Means 군집화 (K=4)
# ==========================================
# 데이터 스케일의 치우침을 완화하기 위해 승하차객수와 매장수에 로그 변환(log1p)을 적용합니다.
districts['승하차객수_log'] = np.log1p(districts['총_승하차객수'])
districts['매장수_log'] = np.log1p(districts['전체_매장수'])

# 군집화 모델의 입력으로 사용할 핵심 피처 컬럼들을 지정합니다.
target_cols = ['승하차객수_log', '매장수_log', '가성비_지수', '프리미엄_지수']

# 데이터 피처들의 단위를 맞춰주기 위해 표준 정규화(StandardScaler)를 적용합니다.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(districts[target_cols].fillna(0))

# 가중치 배열을 곱하여 가성비 및 프리미엄 평가지표의 군집 반영 영향력을 각각 4배로 강화합니다.
X_weighted = X_scaled * np.array([1.0, 1.0, 4.0, 4.0]) 

# 설정된 클래스 수(K=4)로 K-Means 군집화를 수행하고 각 상권에 클러스터 번호를 부여합니다.
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=50)
districts['클러스터'] = kmeans.fit_predict(X_weighted)

# 실루엣 점수 계산 및 출력
sil_avg = silhouette_score(X_weighted, districts['클러스터'])
print(f"✅ 분석 완료 (K={n_clusters}, 실루엣 점수: {sil_avg:.4f})")

# ==========================================
# 5. 결과 시각화 및 리포트
# ==========================================
# 분석 결과를 파악하기 위한 산점도 그래프를 시각화합니다.
plt.figure(figsize=(14, 8))
sns.scatterplot(
    data=districts, 
    x='가성비_지수', 
    y='프리미엄_지수', 
    hue='클러스터', 
    size='총_승하차객수',
    sizes=(60, 600),
    palette='Set1',
    alpha=0.6
)
plt.title(f'가성비 & 프리미엄 종합 점수 기반 상권 클러스터링 (K={n_clusters})')
plt.xlabel('총 가성비 점수 합계')
plt.ylabel('총 프리미엄 점수 합계')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()

# 지정된 6개 필수 컬럼을 추출하고, 가성비 지수와 프리미엄 지수의 합이 높은 상권 순으로 정렬합니다.
districts['종합_지수'] = districts['가성비_지수'] + districts['프리미엄_지수']
final_districts = districts[[
    '상권_ID', '역명', '지역', '가성비_지수', '프리미엄_지수', '클러스터'
]].copy()

# 정렬 기준은 데이터 유효성 검증용으로 종합 지수 기준 내림차순 정렬을 유지합니다.
final_districts['temp_sort'] = districts['종합_지수']
final_districts = final_districts.sort_values(by='temp_sort', ascending=False).drop(columns=['temp_sort'])

# 상위 20개 리포트를 콘솔 창에 출력하여 확인합니다.
print("\n📊 [종합 점수 높은 순 Top 20 결과 확인]")
print(final_districts.head(20))

# ==========================================
# 6. 결과 저장 및 파일 생성
# ==========================================
# 가성비 & 프리미엄 위주로 정렬 및 필터링된 통합본 CSV 저장
merged_file_path = 'OPTICS_상권분석_통합.csv'
final_districts.to_csv(merged_file_path, index=False, encoding='utf-8-sig')
print(f"\n✅ 전체 통합 분석 결과물 저장 완료: {merged_file_path}")

# 역명/지역명 요약본 저장 (기존 포맷 유지) 코드 복원
# station_col = '역명' if '역명' in final_districts.columns else '지하철역'
# region_col = '지역' if '지역' in final_districts.columns else '자치구명'

# if station_col in final_districts.columns and region_col in final_districts.columns:
#     simple_districts = final_districts[[station_col, region_col, '클러스터']].copy()
#     simple_districts = simple_districts.sort_values(by=['클러스터', station_col])
    
#     output_cols = [station_col, region_col]
#     final_csv = simple_districts[output_cols]
    
#     group_file_path = 'DBSCAN_역별_상권묶음_통합(서울, 경기).csv'
#     final_csv.to_csv(group_file_path, index=False, encoding='utf-8-sig')
#     print(f"✅ 역명/지역명 통합 상권 묶음 파일 저장 완료: {group_file_path}")
```
